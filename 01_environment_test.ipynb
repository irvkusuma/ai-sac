```python
import pandas as pd
import numpy as np

data = {
    "Model": ["Actual", "Budget", "Forecast"],
    "Revenue": [150000000, 180000000, 165000000],
    "Cost": [90000000, 100000000, 95000000],
}

df = pd.DataFrame(data)

df["Profit"] = df["Revenue"] - df["Cost"]
df["Margin"] = df["Profit"] / df["Revenue"] * 100

display(df)
```
